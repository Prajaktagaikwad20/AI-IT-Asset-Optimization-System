```python
import pandas as pd
import matplotlib.pyplot as plt


# ---------------------------
# Clean Asset Data
# ---------------------------
asset = pd.DataFrame({
    "Status": ["Assigned", "Maintenance"],
    "Count": [3, 3]
})


# ---------------------------
# Clean Maintenance Data
# ---------------------------
maintenance = pd.DataFrame({
    "Asset": [
        "Dell PowerEdge R740",
        "Lenovo ThinkCentre M90",
        "Dell Latitude 5420"
    ],
    "Cost": [
        25000,
        20000,
        8000
    ]
})


# ---------------------------
# Clean Ticket Data
# ---------------------------
tickets = pd.DataFrame({
    "Status": [
        "Resolved",
        "Open",
        "Pending"
    ],
    "Count": [
        3,
        1,
        1
    ]
})


# ---------------------------
# Clean License Data
# ---------------------------
license_data = pd.DataFrame({
    "Category": [
        "Used Licenses",
        "Available Licenses"
    ],
    "Count": [
        418,
        102
    ]
})


# ---------------------------
# Create Dashboard
# ---------------------------

fig, ax = plt.subplots(2,2, figsize=(12,8))


# Asset Status
ax[0,0].pie(
    asset["Count"],
    labels=asset["Status"],
    autopct="%1.1f%%"
)
ax[0,0].set_title("IT Asset Status")


# Maintenance Cost
ax[0,1].bar(
    maintenance["Asset"],
    maintenance["Cost"]
)
ax[0,1].set_title("Maintenance Cost Analysis")
ax[0,1].tick_params(axis="x", rotation=45)


# Ticket Status
ax[1,0].pie(
    tickets["Count"],
    labels=tickets["Status"],
    autopct="%1.1f%%"
)
ax[1,0].set_title("Support Ticket Status")


# License Usage
ax[1,1].bar(
    license_data["Category"],
    license_data["Count"]
)
ax[1,1].set_title("Software License Utilization")


plt.suptitle(
    "AI-Assisted IT Asset Optimization Dashboard",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    "IT_Asset_Dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
