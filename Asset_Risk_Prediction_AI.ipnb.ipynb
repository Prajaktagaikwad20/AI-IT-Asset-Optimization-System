```python
import pandas as pd

# Create dataset from IT asset history

data = {
    "Asset_Age": [4, 3, 2, 1, 5, 2],
    "Repair_Count": [1, 2, 2, 0, 3, 1],
    "Maintenance_Cost": [25000, 20000, 8000, 0, 35000, 5000],
    "Asset_Value": [300000, 55000, 65000, 70000, 120000, 50000],
    "Risk_Level": [
        "High",
        "High",
        "Medium",
        "Low",
        "High",
        "Low"
    ]
}

df = pd.DataFrame(data)

display(df)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Risk_Level"] = encoder.fit_transform(df["Risk_Level"])

display(df)
```

```python
from sklearn.model_selection import train_test_split

# Input features
X = df[
    [
        "Asset_Age",
        "Repair_Count",
        "Maintenance_Cost",
        "Asset_Value"
    ]
]

# Target
y = df["Risk_Level"]


# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

# Create model
model = DecisionTreeClassifier(
    random_state=42
)

# Train model
model.fit(
    X_train,
    y_train
)

print("Model trained successfully!")
```

```python
new_asset = [[
    4,       # Asset age
    3,       # Repair count
    30000,   # Maintenance cost
    60000    # Asset value
]]

prediction = model.predict(new_asset)

print("Predicted Risk Level:", prediction)
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Predict test data
y_pred = model.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)

# Detailed report
print(classification_report(y_test, y_pred))
```

```python
risk_mapping = {
    0: "Low Risk",
    1: "Medium Risk",
    2: "High Risk"
}

prediction = model.predict([[4,3,30000,60000]])

risk = risk_mapping[prediction[0]]

print("Asset Risk Prediction:", risk)
```

```python
import pickle

with open("asset_risk_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("Model saved successfully!")
```
